```python
import mne
from mne.datasets import eegbci

archivos = eegbci.load_data(subjects=1, runs=[4])
raw = mne.io.read_raw_edf(archivos[0], preload=True)
print(raw.info)
```

```python
mne.set_log_level('WARNING')  # para que no tire tanto texto de debug
raw.plot(duration=5, n_channels=10)
```
